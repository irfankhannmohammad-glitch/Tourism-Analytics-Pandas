```python
# ============================================
# TOURISM ANALYTICS - MAIN ANALYSIS
# ============================================

# 1. Create 2019-2020 comparison dataset
comparison = df[
    ["Country Name", "Country Code", 2019.0, 2020.0]
].dropna(subset=[2019.0, 2020.0]).copy()

# Remove known aggregate/region rows
exclude = [
    "IDA & IBRD total",
    "North America",
    "IDA only"
]

comparison = comparison[
    ~comparison["Country Name"].isin(exclude)
]

# 2. Calculate absolute and percentage change
comparison["Change"] = (
    comparison[2020.0] - comparison[2019.0]
)

comparison["Change (%)"] = (
    comparison["Change"] / comparison[2019.0] * 100
)

# 3. Calculate recovery percentage
comparison["Recovery (%)"] = (
    comparison[2020.0] / comparison[2019.0] * 100
)

# ============================================
# TOP 10 COUNTRIES - 2019
# ============================================

top_10_2019 = (
    countries_2019
    .sort_values(by=2019.0, ascending=False)
    .head(10)
    .copy()
)

print("========== TOP 10 TOURISM DESTINATIONS - 2019 ==========")
display(top_10_2019)

# ============================================
# BIGGEST TOURISM DECLINES
# ============================================

biggest_declines = (
    comparison
    .sort_values(by="Change (%)", ascending=True)
    .head(10)
)

print("\n========== BIGGEST DECLINES: 2019 → 2020 ==========")
display(biggest_declines)

# ============================================
# BEST 2020 RECOVERY RELATIVE TO 2019
# ============================================

best_recovery = (
    comparison
    .sort_values(by="Recovery (%)", ascending=False)
    .head(10)
)

print("\n========== HIGHEST 2020 RECOVERY ==========")
display(best_recovery)

# ============================================
# OVERALL TOURISM CHANGE
# ============================================

total_2019 = comparison[2019.0].sum()
total_2020 = comparison[2020.0].sum()

overall_change = total_2020 - total_2019
overall_change_pct = (overall_change / total_2019) * 100

print("\n========== OVERALL TOURISM CHANGE ==========")
print(f"2019 total arrivals: {total_2019:,.0f}")
print(f"2020 total arrivals: {total_2020:,.0f}")
print(f"Change: {overall_change:,.0f}")
print(f"Percentage change: {overall_change_pct:.2f}%")

# ============================================
# SUMMARY
# ============================================

print("\n========== PROJECT SUMMARY ==========")
print(f"Countries with 2019 & 2020 data: {len(comparison)}")
print(f"Top destination in 2019: {top_10_2019.iloc[0]['Country Name']}")
print(
    f"Largest percentage decline: "
    f"{biggest_declines.iloc[0]['Country Name']} "
    f"({biggest_declines.iloc[0]['Change (%)']:.2f}%)"
)
print(
    f"Highest 2020 recovery: "
    f"{best_recovery.iloc[0]['Country Name']} "
    f"({best_recovery.iloc[0]['Recovery (%)']:.2f}% of 2019)"
)

# ============================================
# SAVE ANALYSIS DATA
# ============================================

comparison.to_excel(
    "Tourism_2019_2020_Analysis.xlsx",
    index=False
)

top_10_2019.to_excel(
    "Top_10_Tourism_2019.xlsx",
    index=False
)

print("\n✅ Analysis completed!")
print("✅ Excel files created successfully!")
```

```python
# ============================================
# TOURISM OPPORTUNITY SCORE
# ============================================

opportunity = comparison[
    ["Country Name", "Country Code",
     2019.0, 2020.0,
     "Change (%)", "Recovery (%)"]
].copy()

# Normalize 2019 arrivals
opportunity["Arrival Score"] = (
    opportunity[2019.0] /
    opportunity[2019.0].max() * 100
)

# Normalize recovery
opportunity["Recovery Score"] = (
    opportunity["Recovery (%)"] /
    opportunity["Recovery (%)"].max() * 100
)

# Score based on tourism demand + recovery
opportunity["Opportunity Score"] = (
    0.60 * opportunity["Arrival Score"] +
    0.40 * opportunity["Recovery Score"]
)

# Rank countries
opportunity["Rank"] = (
    opportunity["Opportunity Score"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# Show top 15
top_opportunities = (
    opportunity
    .sort_values("Opportunity Score", ascending=False)
    .head(15)
)

top_opportunities[
    ["Rank", "Country Name", "Country Code",
     2019.0, 2020.0,
     "Change (%)", "Recovery (%)",
     "Opportunity Score"]
]
```
